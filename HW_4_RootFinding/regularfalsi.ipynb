```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
##Function 1: F(X) = 1
def f(x):
    return np.ones_like(x)

x = np.linspace(-6, 6, 100)
plt.figure()
plt.plot(x, f(x), color='lightgreen', lw=5, label="f(x) = 2x+1")
plt.xlim(-6, 6)
plt.ylim(-10, 10)
plt.axhline(0, color='black')
plt.axvline(0, color='black')

a = -6
b = 6
xvalues = []
cols = ['r', 'b', 'g']

if not f(a) * f(b) < 0:    
    print("no sign change")
else:
    for i in range(3):                              
        c = round((a*f(b) - b*f(a)) / (f(b) - f(a)), 2)      
        xvalues.append(c)
        plt.scatter(c, 0, s=200, color=cols[i], zorder=3, label=f"x{i+1} = {c}")
        if f(c) == 0:        
            break
        if f(a) * f(c) < 0:  
            b = c
        else:               
            a = c

print(xvalues)
plt.legend(loc="best")
plt.title("f(x) = 1")
plt.show()
```
