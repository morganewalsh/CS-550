```python
import numpy as np 
import matplotlib.pyplot as plt 
import sympy as sp
```

```python
##Functions & Derivatives

FXDX = [
    ("1", lambda x: 1.0 + 0*x, lambda x: 0.0 + 0*x),
    ("1.5x", lambda x: 1.5*x, lambda x: 1.5 + 0*x),
    ("1.5x^2", lambda x: 1.5*x**2, lambda x: 3.0*x),
    ("1.5x^3", lambda x: 1.5*x**3, lambda x: 4.5*x**2),
    ("1.5sqrt(x)", lambda x: 1.5*np.sqrt(x), lambda x: 0.75/np.sqrt(x)),
    ("1.5e^x", lambda x: 1.5*np.exp(x), lambda x: 1.5*np.exp(x)),
    ("2x+1", lambda x: 2*x + 1, lambda x: 2.0 + 0*x),
    ("2log(x)", lambda x: 2*np.log(x), lambda x: 2/x),
    ("2sin(x)", lambda x: 2*np.sin(x), lambda x: 2*np.cos(x)),
    ("2cos(x)", lambda x: 2*np.cos(x), lambda x: -2*np.sin(x)),
]
```

```python
#Function 1; fx = 1 
def f(x):
    return np.ones_like(x)

def newtondf(a):
    return (f(a + 0.00001) - f(a)) / 0.00001

def tangent(a):
    return newtondf(a) * (x - a) + f(a)

def n(a):
    return round(a - f(a) / newtondf(a), 2)
x = np.linspace(-6, 6, 100)

x0 = 2
x1 = n(x0)
x2 = n(x1)
x3 = n(x2)
print(x0, x1, x2, x3)

ax = plt.figure()
plt.plot(x, f(x), color='green', lw=6)   
plt.plot(x, tangent(x0), color='red', lw = 3)               
plt.plot(x, tangent(x1), color='blue', lw = 3)              
plt.plot(x, tangent(x2), color='darkgreen', lw = 3)
plt.xlim(-6, 6)
plt.ylim(-10, 15)
plt.axhline(color='black')
plt.axvline(color='black')

plt.legend(["f(x) = 1.5x", "tangent 1", "tangent 2", "tangent 3"])
plt.title("f(x) = 1", loc='center', fontsize=18)
plt.show()
```
