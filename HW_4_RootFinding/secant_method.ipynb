```python
import numpy as np 
import matplotlib.pyplot as plt
import sympy as sp
```

```python
##Function 1 f(x) =1 

def f(x):
    return np.ones_like(x)

x = np.linspace(-6, 6, 1000)


def lineforsecant(a, b):
    if a == b:                       
        return np.full_like(x, np.nan)
    return (f(b) - f(a)) / (b - a) * (x - b) + f(b)

def secant(a, b):
    if f(b) == 0:                     
        return b
    if f(b) - f(a) == 0:             
        return np.nan
    return round(b - f(b) * (b - a) / (f(b) - f(a)), 2)

s0 = 2
s1 = 2.5
s2 = secant(s0, s1)
s3 = secant(s1, s2)
s4 = secant(s2, s3)
print(s0, s1, s2, s3, s4)

ax = plt.figure()
plt.plot(x, f(x), color='lightgreen', lw=3)
plt.plot(x, lineforsecant(s0, s1), color='red', lw=3)
plt.plot(x, lineforsecant(s1, s2), color='blue', lw=3)
plt.plot(x, lineforsecant(s2, s3), color='darkgreen', lw=3, ls='--')
plt.xlim(-6, 6)
plt.ylim(-10, 10)
plt.axhline(color='black')
plt.axvline(color='black')
plt.plot(s0, 0, 'o', color='black')
plt.plot(s1, 0, 'o', color='black')
plt.plot(s2, 0, 'o', color='black')
plt.plot(s3, 0, 'o', color='black')
plt.plot(s4, 0, 'o', color='black')
plt.legend(["f(x) = 1", "secant 1", "secant 2", "secant 3"])
plt.title("f(x) = 1", loc='center', fontsize=18)
plt.show()
```

```python
##Function 1: F(X) = 1 

def f(x):
    return np.ones_like(x)

x = np.linspace(-6, 6, 1000)
plt.figure()
plt.plot(x, f(x), color='lightgreen', lw=5, label="f(x) = 1")
plt.xlim(-6, 6)
plt.ylim(-10, 10)
plt.axhline(0, color='black')
plt.axvline(0, color='black')


t = np.arange(-6, 6, 0.1)
a = 2                       
b = 2.5                 

plt.scatter(a, f(a), s=60, color="r")
plt.scatter(b, f(b), s=60, color="r")
plt.plot([a, a], [0, f(a)], ls="--", color="k")
plt.plot([b, b], [0, f(b)], ls="--", color="k")

plt.title("f(x) = 1")

xvalues = []
xvalues.append(a)
xvalues.append(b)
notconverge = 1
count = 0
cols = ['r', 'b', 'g--']    

while (notconverge == 1 and count < 3):
    if f(xvalues[count+1]) - f(xvalues[count]) == 0:    
        print("flat secant")
        break
    slope = (f(xvalues[count+1]) - f(xvalues[count])) / (xvalues[count+1] - xvalues[count])
    intercept = -slope * xvalues[count+1] + f(xvalues[count+1])
    plt.plot(t, slope*t + intercept, cols[count], label=f"secant{count+1}")
    nextval = round(-intercept / slope, 2)          
    xvalues.append(nextval)
    if abs(f(nextval)) < 0.001:
        notconverge = 0
    count = count + 1

print(xvalues)
plt.scatter(xvalues, [0]*len(xvalues), s=60, color='k', zorder=3)
plt.legend(loc = "best")

plt.show()
```
